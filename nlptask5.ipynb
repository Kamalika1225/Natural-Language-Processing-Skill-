```python
import pandas as pd
from nltk.metrics.distance import edit_distance

# -----------------------------------------
# 1. Spelling Dataset
# -----------------------------------------

data = {
    "incorrect": [
        "recieve",
        "teh",
        "freind",
        "becuase",
        "adress",
        "definately",
        "seperate",
        "occured",
        "enviroment",
        "goverment",
        "sucess",
        "neccessary",
        "beleive",
        "tomorow",
        "langauge"
    ],

    "correct": [
        "receive",
        "the",
        "friend",
        "because",
        "address",
        "definitely",
        "separate",
        "occurred",
        "environment",
        "government",
        "success",
        "necessary",
        "believe",
        "tomorrow",
        "language"
    ]
}

df = pd.DataFrame(data)

# Create spelling dictionary
spelling_dict = dict(zip(df["incorrect"], df["correct"]))


# -----------------------------------------
# 2. Spelling Correction Function
# -----------------------------------------

def correct_word(word):

    word_lower = word.lower()

    # Check dataset
    if word_lower in spelling_dict:
        return spelling_dict[word_lower]

    # Find closest word
    suggestions = []

    for correct in df["correct"]:
        distance = edit_distance(word_lower, correct)
        suggestions.append((distance, correct))

    suggestions.sort()

    return suggestions[0][1]


# -----------------------------------------
# 3. Example
# -----------------------------------------

print("Example:")
print("Input  : recieve")
print("Output : receive")


# -----------------------------------------
# 4. User Input
# -----------------------------------------

print("\nNow try it yourself!")

user_input = input("Enter a word: ")

corrected_word = correct_word(user_input)


# -----------------------------------------
# 5. Result
# -----------------------------------------

print("Corrected word:", corrected_word)
```
